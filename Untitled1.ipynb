```python
import streamlit as st

st.title("🧮 Math Quiz")

st.write("What is 5 × 6?")

answer = st.radio(
    "Choose your answer:",
    ["25", "30", "35"]
)

if st.button("Check Answer"):
    if answer == "30":
        st.success("Correct! 🎉")
    else:
        st.error("Wrong answer. Try again!")
```
